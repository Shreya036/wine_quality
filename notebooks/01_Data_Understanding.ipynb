```python
data_dictionary = pd.DataFrame([
    ("fixed acidity",        "Input",  "g/dm3 (tartaric acid)",   "Non-volatile acids (mainly tartaric) that do not evaporate easily; give the wine its tartness."),
    ("volatile acidity",     "Input",  "g/dm3 (acetic acid)",     "Acetic acid level; too much gives an unpleasant vinegar taste."),
    ("citric acid",          "Input",  "g/dm3",                   "Citric acid; adds freshness and flavour in small amounts."),
    ("residual sugar",       "Input",  "g/dm3",                   "Sugar remaining after fermentation stops; higher values mean a sweeter wine."),
    ("chlorides",            "Input",  "g/dm3 (sodium chloride)", "Amount of salt in the wine."),
    ("free sulfur dioxide",  "Input",  "mg/dm3",                  "SO2 not bound to other molecules; prevents microbial growth and oxidation."),
    ("total sulfur dioxide", "Input",  "mg/dm3",                  "Free plus bound SO2; high levels can be noticed in smell and taste."),
    ("density",              "Input",  "g/cm3",                   "Density of the wine; depends mainly on alcohol and sugar content."),
    ("pH",                   "Input",  "pH scale",                "Acidity/alkalinity; lower pH means a more acidic wine."),
    ("sulphates",            "Input",  "g/dm3 (potassium sulphate)", "Wine additive that acts as an antimicrobial and antioxidant."),
    ("alcohol",              "Input",  "% by volume",             "Alcohol content of the wine."),
    ("quality",              "Target", "score (0-10)",            "Sensory quality score given by wine tasters (median of at least 3 evaluations)."),
], columns=["Variable", "Role", "Unit", "Meaning"])

data_dictionary["Observed min"] = data_dictionary["Variable"].map(df.min())
data_dictionary["Observed max"] = data_dictionary["Variable"].map(df.max())

import os
os.makedirs("../outputs", exist_ok=True)
data_dictionary.to_csv("../outputs/data_dictionary.csv", index=False)
data_dictionary
```

Both
1. Regression: quality is a numerical score (observed range 3 to 9), so we can predict it directly from the 11 chemical properties (Tasks 5 and 6).
2. Classification: we can group the scores into classes using the project rule Good = quality >= 7, Not Good = quality < 7 and predict the class (Task 7).

Two things to keep in mind:
quality is an integer score (an ordinal rating), so a regression model will predict decimals that are only an approximation of the rating.
The scores are concentrated around 5, 6 and 7, so the Good / Not Good classes will be imbalanced (checked in notebook 06).
Conclusion
The dataset has 4,898 white-wine records, 11 numerical chemical inputs and one numerical target (quality), with no categorical columns. It is suitable for both regression and classification. Duplicate rows and outliers are examined in notebook 02.
